# 4.AttributeError                            

```python

from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Sample data
data = [["Hello world"], ["Machine learning is fun"]]

# Create a CountVectorizer instance
cv = CountVectorizer()

# Fit and transform the data
X = cv.fit_transform(data)

# Convert to DataFrame for better visualization
df = pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out())

# Print the feature names and the resulting matrix
print("Feature Names:", cv.get_feature_names_out())
print("\nCount Vectorized Matrix:\n")
print(df)

# need to make the input a list of strings 
```
