# 1. Topic Modeling using LDA
# Task: Topic Modeling (LDA)

```python
documents = [
    "Cats and dogs are lovely animals.",
    "I love my pet dog.",
    "Programming in Python is fun.",
    "Data science and machine learning with Python.",
    "Cats often sleep during the day.",
    "Machine learning is used in many applications."
]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

documents = [
    "Cats and dogs are lovely animals.",
    "I love my pet dog.",
    "Programming in Python is fun.",
    "Data science and machine learning with Python.",
    "Cats often sleep during the day.",
    "Machine learning is used in many applications."
]

# Convert documents to document-term matrix
vectorizer = CountVectorizer(stop_words='english')
doc_term_matrix = vectorizer.fit_transform(documents)

# Apply LDA with 2 topics
lda = LatentDirichletAllocation(n_components=2, random_state=42)
lda.fit(doc_term_matrix)

# Display topics and their top words
feature_names = vectorizer.get_feature_names_out()
for topic_idx, topic in enumerate(lda.components_):
    top_words_idx = topic.argsort()[-5:][::-1]
    top_words = [feature_names[i] for i in top_words_idx]
    print(f"Topic {topic_idx + 1}: {', '.join(top_words)}")

# Show document-topic distribution
doc_topic_dist = lda.transform(doc_term_matrix)
print("\nDocument-Topic Distribution:")
for i, dist in enumerate(doc_topic_dist):
    print(f"Document {i + 1}: Topic 1={dist[0]:.2f}, Topic 2={dist[1]:.2f}")
```

# 2. How do you convert text to features using TF-IDF?
# Task: Transform a list of sentences into TF-IDF feature vectors.

```python
corpus = [
    "Machine learning is fascinating.",
    "Natural language processing is a part of AI.",
    "AI and machine learning are related fields."
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "Machine learning is fascinating.",
    "Natural language processing is a part of AI.",
    "AI and machine learning are related fields."
]

# Create TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Transform corpus into TF-IDF features
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# Display feature names
feature_names = tfidf_vectorizer.get_feature_names_out()
print("Feature Names:", feature_names)

# Display TF-IDF matrix
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

# Show TF-IDF scores for first document
print("\nTF-IDF scores for first document:")
for word, score in zip(feature_names, tfidf_matrix.toarray()[0]):
    if score > 0:
        print(f"{word}: {score:.4f}")
```

# 3. How do you perform sentiment analysis on text data?
# Task: Analyze the sentiment of a review using VADER.

```python
!pip install vaderSentiment

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize VADER sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Sample review
review = "The product is amazing! I absolutely love it. Best purchase ever."

# Get sentiment scores
sentiment_scores = analyzer.polarity_scores(review)

print("Review:", review)
print("\nSentiment Scores:")
print(f"Positive: {sentiment_scores['pos']:.2f}")
print(f"Neutral: {sentiment_scores['neu']:.2f}")
print(f"Negative: {sentiment_scores['neg']:.2f}")
print(f"Compound: {sentiment_scores['compound']:.2f}")

# Determine overall sentiment
if sentiment_scores['compound'] >= 0.05:
    print("\nOverall Sentiment: POSITIVE")
elif sentiment_scores['compound'] <= -0.05:
    print("\nOverall Sentiment: NEGATIVE")
else:
    print("\nOverall Sentiment: NEUTRAL")
```

# 4. How do you perform Named Entity Recognition (NER)?
# Task: Extract named entities from text using SpaCy.

```python

import spacy

# Load SpaCy English model
nlp = spacy.load("en_core_web_sm")

# Sample text
text = "Apple Inc. was founded by Steve Jobs in Cupertino, California. Elon Musk is the CEO of Tesla."

# Process text
doc = nlp(text)

# Extract named entities
print("Named Entities:\n")
for ent in doc.ents:
    print(f"{ent.text:20} -> {ent.label_:15} ({spacy.explain(ent.label_)})")
```

# 5 How do you represent words using Word2Vec embeddings?
# Task: Train Word2Vec on a sample dataset and get vector for a word.

```python
sentences = [
    ["natural", "language", "processing", "is", "fun"],
    ["machine", "learning", "is", "a", "subset", "of", "AI"],
    ["AI", "includes", "deep", "learning", "and", "NLP"]
]
```

```python
from gensim.models import Word2Vec

sentences = [
    ["natural", "language", "processing", "is", "fun"],
    ["machine", "learning", "is", "a", "subset", "of", "AI"],
    ["AI", "includes", "deep", "learning", "and", "NLP"]
]

# Train Word2Vec model
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=2, epochs=100)

# Get vector for a word
word = "learning"
vector = model.wv[word]

print(f"Vector for '{word}':")
print(vector)

# Find similar words
print(f"\nWords similar to '{word}':")
similar_words = model.wv.most_similar(word, topn=3)
for similar_word, similarity in similar_words:
    print(f"{similar_word}: {similarity:.4f}")

# Check vocabulary
print("\nVocabulary:")
print(list(model.wv.index_to_key))
```
